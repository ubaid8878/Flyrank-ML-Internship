## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
# 2) The question: decision, action, and cost of a wrong call

## Research question

Which pages should an SEO/content team review first for a possible content refresh, based on observable search and engagement signals?

### Decision

The decision is which pages should receive limited content-review attention first.

### Unit of analysis

The unit of analysis is one content page.

### Output

The intended output is a ranked list of pages with a priority score and interpretable reason codes explaining why each page was selected.

### Action

An SEO/content reviewer could inspect the highest-ranked pages first and decide whether to refresh, improve, or leave each page unchanged.

### Cost of a wrong call

A false positive can waste an editor's or SEO specialist's time on a page that does not need attention. A false negative can cause a potentially valuable or declining page to be missed. Therefore, the ranking should prioritize pages where there is enough evidence of demand, weakness, or opportunity.

### Why data or ML can help

A simple rule can identify obvious cases, but many observable signals can interact: impressions, sessions, CTR, average position, content age, freshness, engagement, and other page characteristics. A model may help rank pages when these signals interact in ways that are difficult to capture with a small hand-written rule. However, ML should only be used if it improves the decision compared with a transparent baseline.
```

```python
import pandas as pd

DATA_PATH = "/content/Flyrank-ML-Internship/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(DATA_PATH)

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nTrend direction:")
print(df["trend_direction"].value_counts(dropna=False))

print("\nAverage 90-day impressions:",
      round(df["impressions_90d"].mean(), 2))

print("Average 90-day sessions:",
      round(df["sessions_90d"].mean(), 2))

print("Pages with declining trend:",
      int((df["trend_direction"] == "down").sum()))
```

```python
### What these numbers suggest

The starter dataset gives us a substantial page-level sample and contains measurable search and engagement signals. In particular, the presence of impressions, sessions, content age, and trend direction means that pages can be compared using observable evidence rather than relying only on subjective judgment.

The number of pages with a downward trend is useful evidence that there are pages worth investigating for possible refresh opportunities. These numbers are only descriptive evidence at this stage; they do not prove that a refresh would cause a page to improve.
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

```python
# 4) Careful words: what I can and can't claim

## What I can claim

I can investigate whether observable search and engagement signals are useful for ranking pages for human review. I can compare a simple baseline with a learned method and measure whether the ranking identifies more relevant pages in the available evaluation data.

## What I cannot claim

I cannot claim that the model will guarantee which page should be refreshed. I also cannot claim that refreshing a page will cause its rankings, traffic, or engagement to improve. The starter label is based on the observed trend direction and is a proxy for the underlying business question, not a causal outcome.

The results should therefore be described as observed, directional, and decision-support evidence rather than as proof that the model predicts Google's ranking algorithm or guarantees future page performance.
```
